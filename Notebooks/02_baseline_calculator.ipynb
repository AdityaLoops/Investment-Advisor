# Phase 2: Baseline Compound-Interest Calculator

Pure math baseline (no ML) using A = P(1+r)^t. Rates are placeholder 
constants per fund category — NOT derived from real data yet.

Known limitation: since rates are static, the recommended "best" option 
never changes regardless of `years` or `principal` — this is expected 
and will be fixed in Phase 3/4 once rates are computed from real 
historical NAV data.
SINCE TILL WILL ALWAYS GIVE A FIXED RESULT (WHERE TO INVEST) WE HAVE TO MAKE RATES DYNAMIC SO FOR DIFF AMOUNT OF TIME WE'LL ACCTUALLY KNOW WHAT IS BEST 

```python
import numpy as np
p = 10000
t = 5
r = .12

a = p * (1+r)**t
print(a)
```

```python
def compound_interest(p,r,t):
    a = p * (1+r)**t
    return a
print(compound_interest(10000, .12, 5))
```

```python
options = {
    'hdfc_large_cap': 0.14,
    'hdfc_corporate_bond': 0.08,
    'hdfc_hybrid_equity': 0.11,
    'sbi_large_cap': 0.145,
    'icici_large_cap': 0.13,
    'icici_corporate_bond': 0.075
}
p = 10000
t = 10
for name , rate in options.items():
    a = compound_interest(p, rate, t)
    print(f"{name} : Rs{a: ,.2f}")
```

```python
best_option = max(options, key = lambda name: compound_interest(p, options[name], t))
best_result = compound_interest(p, options[best_option], t)

print(f"Best Option = {best_option}")
print(f"Highest Amount = {best_result}")
```

```python
def recommend_best(p, t, opitons):
    results = {}
    for name, rate in opitons.items():
        results[name] = compound_interest(p, rate, t)

    best = max(options, key= results.get)
    return  results, best

result, best = recommend_best(10000, 10, options)
print(result)
print('Winner:', best)
```
